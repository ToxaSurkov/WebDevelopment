## Markdown отображается как содержимое документа

- **Этот текст** не является Python-кодом.
- Формула $E = mc^2$ обрабатывается интерфейсом notebook.
- Сохранение текста в `.ipynb` не требует выполнения Python Kernel.

```python
# Вычисления делает Python Kernel, а браузер показывает rich output (HTML/SVG).
from IPython.display import display, HTML
values = [8, 14, 11, 19]
bars = "".join(
    f'<rect x="{15 + i*60}" y="{125-v*5}" width="40" height="{v*5}" fill="#0e6c73"/>'
    for i, v in enumerate(values)
)
svg = f'<svg width="280" height="145" viewBox="0 0 280 145" xmlns="http://www.w3.org/2000/svg">{bars}</svg>'
display(HTML('<b>График, вычисленный в Kernel:</b><br>' + svg))
```

## Для демонстрации состояния Kernel

1. Сначала выполните первые две кодовые ячейки: `x` существует.
2. В меню **Kernel → Restart Kernel** перезапустите ядро.
3. Повторно выполните следующую ячейку. Возникнет `NameError`.
4. При этом текст, Markdown и ранее сохранённые результаты в `.ipynb` остаются частью документа.
5. Сохраните файл и откройте `.ipynb` в текстовом редакторе: увидите JSON.

```python
print(x)  # После перезапуска Kernel без повторного определения x будет NameError
```
